```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
life_clean= pd.read_csv("Data/clean/life_expectancy_clean.csv")
```

```python
import plotly.graph_objects as go

top_3 = (
    life_clean[life_clean["Year"] == 2015]
    .groupby("Country")["Life expectancy"]
    .mean()
    .nlargest(3)
    .sort_values(ascending=False)
)

fig_top3 = go.Figure()

fig_top3.add_trace(go.Bar(
    x=top_3.index,
    y=top_3.values,
    marker_color="steelblue",
    text=[f"{v:.1f}" for v in top_3.values],
    textposition="outside"
))

fig_top3.update_layout(
    title="Les 3 pays avec la plus grande espérance de vie en 2015",
    xaxis_title="Pays",
    yaxis_title="Espérance de vie (années)",
    yaxis=dict(range=[top_3.min() - 2, top_3.max() + 2]),
    template="plotly_white",
    showlegend=False
)

fig_top3.show()
```

```python
bottom_3 = (
    life_clean[life_clean["Year"] == 2015]
    .groupby("Country")["Life expectancy"]
    .mean()
    .nsmallest(3)
    .sort_values(ascending=True)
)

fig_bottom3 = go.Figure()

fig_bottom3.add_trace(go.Bar(
    x=bottom_3.index,
    y=bottom_3.values,
    marker_color="indianred",
    text=[f"{v:.1f}" for v in bottom_3.values],
    textposition="outside"
))

fig_bottom3.update_layout(
    title="Les 3 pays avec la plus faible espérance de vie en 2015",
    xaxis_title="Pays",
    yaxis_title="Espérance de vie (années)",
    yaxis=dict(range=[bottom_3.min() - 2, bottom_3.max() + 2]),
    template="plotly_white",
    showlegend=False
)

fig_bottom3.show()
```

```python
import plotly.graph_objects as go

def create_life_expectancy_chart(selected_country):

    # Filter data between 2000 and 2015
    df = life_clean[
        life_clean["Year"].between(2000, 2015)
    ].copy()

    # Selected country
    country_data = (
        df[df["Country"] == selected_country]
        .groupby("Year", as_index=False)["Life expectancy"]
        .mean()
        .sort_values("Year")
    )

    # Average life expectancy by development status
    status_data = (
        df.groupby(["Year", "Status"], as_index=False)["Life expectancy"]
        .mean()
    )

    fig = go.Figure()

    # Developed countries
    developed = status_data[
        status_data["Status"] == "Developed"
    ]

    fig.add_trace(go.Scatter(
        x=developed["Year"],
        y=developed["Life expectancy"],
        mode="lines",
        name="Pays développés",
        line=dict(color="seagreen", width=3)
    ))

    # Developing countries
    developing = status_data[
        status_data["Status"] == "Developing"
    ]

    fig.add_trace(go.Scatter(
        x=developing["Year"],
        y=developing["Life expectancy"],
        mode="lines",
        name="Pays en développement",
        line=dict(color="darkorange", width=3)
    ))

    # Selected country
    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["Life expectancy"],
        mode="lines+markers",
        name=selected_country,
        line=dict(color="royalblue", width=4),
        marker=dict(size=7)
    ))

    fig.update_layout(
        title="L'évolution de l'espérance de vie (2000–2015)",
        xaxis_title="Année",
        yaxis_title="Espérance de vie (années)",
        template="plotly_white",
        hovermode="x unified",
        xaxis=dict(dtick=1),
        legend=dict(
            orientation="h",
            y=-0.2,
            x=0.5,
            xanchor="center"
        ),
        margin=dict(t=70, b=100)
    )

    return fig
```

```python
fig = create_life_expectancy_chart("Mexico")
fig.show()
```
